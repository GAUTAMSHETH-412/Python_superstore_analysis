Hypothesis 1: Technology products have the highest profit margin compared to other product categories

```python
cat_profit = df.groupby('Category')['Profit'].sum()
cat_profit.plot(kind='bar')
plt.title('Profit by category')
plt.ylabel('Total')
plt.xlabel('Category')
```

Hyppthesis 2: The east region has the highest sales compared to other regions

```python
df.columns
```

```python
east_sales = df.groupby('Region')['Sales'].sum()
east_sales.plot(kind='bar')
plt.title('Total sales by Region')
plt.ylabel('Total Sales')
# Conclusion: The hypothesis is not supported as the central region has the highest  sales
```

Hypothesis 3: Sales are higher during certain months of the year

```python
df.columns
```

```python
df['Order Month'] = pd.DatetimeIndex(df['Order Date']).month
```

```python
Month_sale = df.groupby('Order Month')['Sales'].sum()
Month_sale.plot(kind='line')
plt.title('Total Sales by Month')
plt.ylabel('Total Sales')
plt.show()
# Conclusion: Our hypothesis is supported as sales are higher during certain months of the year
```

Hypothesis 4: Orders with same-day shipping have the lowest rate of returned products

```python
df.columns
```

```python
total_orders_by_shipping_mode= df.groupby('Ship Mode').size()
total_orders_by_shipping_mode
```

```python
returned_orders_by_shipping_mode = df[df['Profit']<0].groupby('Ship Mode').size()
returned_orders_by_shipping_mode
```

```python
returned_per_by_shipping = (returned_orders_by_shipping_mode / total_orders_by_shipping_mode) * 100
returned_per_by_shipping
```

```python
returned_per_by_shipping.plot(kind='bar')
plt.title('Return Per By Shipping Mode')
plt.ylabel('Return Per')
plt.xlabel('Shipping Mode')
plt.show()
# Conclusion : The hypothesis  is supported as orders with same day shipping have the lowest rate of return products
```

Hypothesis 5: The company's profit is more on weekdays than on weekends.

```python
df.columns
```

```python
df['Order Day']= pd.DatetimeIndex(df['Order Date']).day_name()
```

```python
day_profit = df.groupby('Order Day')['Profit'].sum()
day_profit
```

```python
day_profit.plot(kind='bar')
plt.title('Total profit by the day of the week')
plt.xlabel('Day of the week')
plt.ylabel('Total Profit')
plt.show()
# Conclusion : The hypothesis is supported as company's profit is higher on weekdays than weekends
```
